# SedTRAILS trajectory visualization

This notebook demonstrates reusable trajectory plotting utilities. Bathymetry backgrounds should use the SEAWAD colormap; particle age and source-distance plots use sequential colormaps such as `viridis`.

```python
from pathlib import Path

import matplotlib.pyplot as plt

from sedtrails.pathway_visualizer.trajectories import read_netcdf
from sedtrails.pathway_visualizer.sedtrails_plotting import (
    animate_particles,
    compute_particle_stats,
    load_from_xarray,
    particles_include_exclude,
    plot_trajectories_by_age,
    plot_trajectories_by_baseline,
    quick_explorer,
    stats_to_csv,
)

def resolve_results_file(filename: str = 'sedtrails_results.nc') -> Path:
    candidates = [
        Path.cwd() / 'examples' / 'results' / filename,
        Path.cwd() / '..' / 'results' / filename,
    ]
    for candidate in candidates:
        resolved = candidate.resolve()
        if resolved.exists():
            return resolved
    raise FileNotFoundError(
        f'Could not find {filename} in examples/results relative to {Path.cwd()}'
    )

results_file = resolve_results_file()
ds = read_netcdf(results_file)
tr = load_from_xarray(ds)
tr
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
plot_trajectories_by_age(tr, cmap='viridis', ax=ax)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
plot_trajectories_by_baseline(tr, rotation_deg=0.0, cmap='viridis', ax=ax)
plt.show()
```

```python
fig, anim = animate_particles(tr, color_mode='baseline', cmap='viridis', interval_ms=120)
plt.show()
```

```python
stats = compute_particle_stats(tr)
stats_to_csv(stats, results_file.with_name('particle_statistics.csv'))
```

```python
# Optional interactive explorer; works best with an interactive matplotlib backend.
quick_explorer(tr, rotation_deg=0.0)
```
